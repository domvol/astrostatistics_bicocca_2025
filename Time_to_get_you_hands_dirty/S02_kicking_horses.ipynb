## Kicking horses

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy

deaths = np.array([0, 1, 2, 3, 4])
groups = np.array([109, 65, 22, 3, 1])
```

```python
tot_groups = np.sum(groups)
freq = groups / tot_groups
plt.scatter(deaths, freq, label = 'Frequency')
plt.grid()
plt.legend()
plt.title('Frequency vs deaths by kicking horses')
plt.show
```

Normalization is applied to obtain a distribution that represents the probability of observing a certain number of deaths, making the result more general and comparable. This approach allows for meaningful comparisons between datasets of different sizes and enables direct comparison with theoretical probability distributions.

```python
plt.scatter(deaths, freq)
mu_span = np.linspace(0.4, 1, 5)

for mu in mu_span:
    if mu == 0:
        pass
    else:
        poisson = scipy.stats.poisson(mu).pmf(deaths)
        plt.plot(deaths, poisson, label='$\mu='+str(round(mu, 2))+"$")
        plt.legend()
```

```python
mean_mu = np.average(deaths,weights=freq)
me_mu = np.average(deaths)     # just a test
fit_y = scipy.stats.poisson(mean_mu).pmf(deaths)
fit_y2 = scipy.stats.poisson(me_mu).pmf(deaths)  # just a test
print(f'The mean_mu is: {round(mean_mu, 2)}')

# Plot
plt.scatter(deaths, freq, label='Data')
plt.plot(deaths, fit_y, color='red', label='Poisson Fit (weighted mean)')
plt.plot(deaths, fit_y2, color='green', label='Poisson Fit (arithmetic mean)')  # just a test
plt.title('Poisson Fit to Deaths by Kicking Horses')
plt.grid()
plt.legend()
plt.show()
```

The weighted mean provides a more accurate estimate of the central parameter when data are grouped by frequency, because it better represents the structure of the observed data (frequencies are different for each groups).
